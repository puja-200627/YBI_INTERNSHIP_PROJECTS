# Day 10 - Handling Missing Values

In this step, we will handle missing values in the student attendance dataset using Pandas.

Missing values can affect data analysis and calculations. We will check the dataset and handle missing values wherever necessary to maintain data quality and accuracy.

##Checking missing values

```python
import pandas as pd
attendance_data = {
    "Student_ID": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110
    ],

    "Date": [
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02"
    ],

    "Subject": [
        "Python", "Python", "Python", "Python", "Python",
        "Python", "Python", "Python", "Python", "Python",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS"
    ],

    "Status": [
        "Present", "Present", "Absent", "Present", "Present",
        "Absent", "Present", "Present", "Absent", "Present",
        "Present", "Absent", "Present", "Present", "Absent",
        "Present", "Present", "Absent", "Present", "Present"
    ]
}

attendance = pd.DataFrame(attendance_data)
print("Missing Values Before Handling:")
print(attendance.isnull().sum())

print("\nTotal Missing Values:",
      attendance.isnull().sum().sum())
```

##Handling missing values

```python
if attendance.isnull().sum().sum() == 0:
    print("No missing values found. Dataset is already clean.")
else:
    attendance["Name"] = attendance["Name"].fillna("Unknown")
    attendance["Branch"] = attendance["Branch"].fillna("Unknown")
    attendance["Section"] = attendance["Section"].fillna("Unknown")
    attendance["Subject"] = attendance["Subject"].fillna("Unknown")
    attendance["Status"] = attendance["Status"].fillna("Absent")

    print("Missing values handled successfully.")
```

##Verifying the dataset

```python
print("Missing Values After Handling:")
print(attendance.isnull().sum())

print("\nTotal Missing Values After Handling:",
      attendance.isnull().sum().sum())
```

## Conclusion

In this step, missing values in the Student Attendance Management System dataset were checked and handled using the Pandas library in Python.

The `isnull()` function was used to identify missing values, and the `sum()` function was used to calculate the number of missing values in each column. An `if-else` condition was implemented to determine whether missing values existed in the dataset.

If missing values are found, the `fillna()` function replaces them with suitable values. For text-based columns such as Name, Branch, Section, and Subject, the missing entries are replaced with "Unknown". Missing values in the Status column are replaced with "Absent" for this demonstration.

The current dataset contains no missing values, so the program confirms that the dataset is already clean and does not make unnecessary changes.

Handling missing values is an important data preprocessing step because incomplete records can affect the accuracy of analysis and reports. This process helps maintain data quality and prepares the dataset for further operations.

Overall, this step improved our understanding of missing-value treatment and prepared the attendance dataset for the next stage of data cleaning, which involves identifying duplicate records.

# Day 17 - Creating Attendance Status

In this step, we will classify students according to their attendance percentage.

Students with attendance of 75% or above will be classified as "Good Attendance", while students below 75% will be classified as "Low Attendance". ##Creating attendance status

```python
total_classes = attendance.groupby("Student_ID").size()

present_classes = (
    attendance[attendance["Status"] == "Present"]
    .groupby("Student_ID")
    .size()
)

attendance_summary = pd.DataFrame({
    "Total_Classes": total_classes,
    "Present_Classes": present_classes
}).fillna(0)

attendance_summary["Present_Classes"] = (
    attendance_summary["Present_Classes"].astype(int)
)

attendance_summary = attendance_summary.reset_index()
attendance_summary["Attendance_Percentage"] = (
    attendance_summary["Present_Classes"]
    / attendance_summary["Total_Classes"]
    * 100
).round(2)
attendance_summary["Attendance_Status"] = (
    attendance_summary["Attendance_Percentage"].apply(
        lambda percentage: "Good Attendance"
        if percentage >= 75
        else "Low Attendance"
    )
)

display(attendance_summary)
```

# Day 24: Basic Statistical Analysis of Student Attendance

## Objective
The objective of Day 24 is to perform basic statistical analysis on student attendance data to understand overall attendance patterns.

## Methodology
1. Calculate the total number of attendance records.
2. Count the total present and absent records.
3. Calculate the overall attendance percentage.
4. Find the average, maximum, and minimum student attendance percentages.
5. Display the statistical results.

## Expected Outcome
A statistical report showing the total attendance records, present and absent counts, overall attendance percentage, and summary statistics of student attendance.

##Calculating Basic Statistics

```python
import pandas as pd
attendance_data = {
    "Student_ID": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110
    ],

    "Date": [
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02"
    ],

    "Subject": [
        "Python", "Python", "Python", "Python", "Python",
        "Python", "Python", "Python", "Python", "Python",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS"
    ],

    "Status": [
        "Present", "Present", "Absent", "Present", "Present",
        "Absent", "Present", "Present", "Absent", "Present",
        "Present", "Absent", "Present", "Present", "Absent",
        "Present", "Present", "Absent", "Present", "Present"
    ]
}

attendance = pd.DataFrame(attendance_data)
# Calculate total attendance records
total_records = len(attendance_data)

# Count present and absent records
present_records = (
    attendance["Status"] == "Present"
).sum()

absent_records = (
    attendance["Status"] == "Absent"
).sum()

# Calculate overall attendance percentage
if total_records > 0:
    overall_percentage = (
        present_records / total_records
    ) * 100
else:
    overall_percentage = 0

# Calculate attendance percentage for each student
total_classes = attendance.groupby("Student_ID").size()

present_classes = (
    attendance[attendance["Status"] == "Present"]
    .groupby("Student_ID")
    .size()
)

student_percentage = (
    present_classes.reindex(total_classes.index, fill_value=0)
    / total_classes
) * 100

# Calculate statistical values
average_attendance = student_percentage.mean()
maximum_attendance = student_percentage.max()
minimum_attendance = student_percentage.min()

# Display the results
print("BASIC ATTENDANCE STATISTICAL REPORT")
print("-" * 40)
print("Total Attendance Records:", total_records)
print("Present Records:", present_records)
print("Absent Records:", absent_records)
print("Overall Attendance Percentage:", round(overall_percentage, 2), "%")
print("Average Student Attendance:", round(average_attendance, 2), "%")
print("Maximum Student Attendance:", round(maximum_attendance, 2), "%")
print("Minimum Student Attendance:", round(minimum_attendance, 2), "%")
```

| `.mean()` | Calculates average student attendance. |
| `.max()` | Finds the highest attendance percentage. |

## Methodology
1. Convert attendance dates into datetime format.

3. Group attendance records by month.
4. Calculate total, present, and absent records.
5. Calculate monthly attendance percentages and visualize the results.

## Expected Outcome
A monthly attendance report and a bar chart showing attendance percentages for each month.

##Preparing Monthly Attendance Data

```python
import pandas as pd
import matplotlib.pyplot as plt

attendance_data = {
    "Student_ID": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110
    ],

    "Date": [
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02"
    ],

    "Subject": [
        "Python", "Python", "Python", "Python", "Python",
        "Python", "Python", "Python", "Python", "Python",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS"
    ],

    "Status": [
        "Present", "Present", "Absent", "Present", "Present",
        "Absent", "Present", "Present", "Absent", "Present",
        "Present", "Absent", "Present", "Present", "Absent",
        "Present", "Present", "Absent", "Present", "Present"
    ]
}

attendance = pd.DataFrame(attendance_data)

# Work with a copy of the attendance dataset
monthly_data = attendance.copy()

# Convert dates into datetime format
monthly_data["Date"] = pd.to_datetime(
    monthly_data["Date"], errors="coerce"
)

# Remove records with invalid dates
monthly_data = monthly_data.dropna(subset=["Date"])

# Extract month and year
monthly_data["Month"] = (
    monthly_data["Date"].dt.to_period("M").astype(str)
)

# Calculate monthly attendance summary
monthly_summary = monthly_data.groupby("Month").agg(
    Total_Records=("Status", "count"),
    Present_Records=("Status", lambda x: (x == "Present").sum()),
    Absent_Records=("Status", lambda x: (x == "Absent").sum())
).reset_index()

# Calculate attendance percentage
monthly_summary["Attendance_Percentage"] = (
    monthly_summary["Present_Records"]
    / monthly_summary["Total_Records"] * 100
).round(2)

print("Monthly Attendance Analysis:")
display(monthly_summary)
```

For each month, the total number of attendance records, present records, and absent records were calculated. The monthly attendance percentage was then determined by dividing the present records by the total records and multiplying the result by 100.

In the original sample dataset, all 20 attendance records belong to September 2026. Out of these records, 14 are marked as Present and 6 are marked as Absent, resulting in an attendance percentage of 70%.

2. Pseudocode
1. Start.

3. Check for missing values.
4. Remove duplicate records.
5. Correct date and attendance data types.

##Final Data Cleaning

```python

import pandas as pd

# Convert student dictionary into DataFrame
if isinstance(students, dict):
    students = pd.DataFrame(students)

# Convert attendance dictionary into DataFrame
if isinstance(attendance_data, dict):
    attendance_data = pd.DataFrame(attendance_data)

# Create copies of original datasets
clean_students = students.copy()
clean_attendance = attendance_data.copy()

# Remove duplicate student records
clean_students = clean_students.drop_duplicates(
    subset="Student_ID"
)

# Remove duplicate attendance records
clean_attendance = clean_attendance.drop_duplicates()

# Remove records with missing essential values
clean_students = clean_students.dropna(
    subset=["Student_ID", "Name"]
)

clean_attendance = clean_attendance.dropna(
    subset=["Student_ID", "Date", "Subject", "Status"]
)

# Correct data types
clean_students["Student_ID"] = pd.to_numeric(
    clean_students["Student_ID"], errors="coerce"
)

clean_attendance["Student_ID"] = pd.to_numeric(
    clean_attendance["Student_ID"], errors="coerce"
)

clean_attendance["Date"] = pd.to_datetime(
    clean_attendance["Date"], errors="coerce"
)

# Remove invalid records
clean_students = clean_students.dropna(
    subset=["Student_ID"]
)

clean_attendance = clean_attendance.dropna(
    subset=["Student_ID", "Date"]
)

# Standardize attendance status
clean_attendance["Status"] = (
    clean_attendance["Status"].astype(str).str.strip().str.title()
)

# Keep only valid attendance statuses
clean_attendance = clean_attendance[
    clean_attendance["Status"].isin(["Present", "Absent"])
]

# Keep records of registered students only
valid_student_ids = clean_students["Student_ID"].unique()

clean_attendance = clean_attendance[
    clean_attendance["Student_ID"].isin(valid_student_ids)
]

# Reset indexes
clean_students = clean_students.reset_index(drop=True)
clean_attendance = clean_attendance.reset_index(drop=True)

# Display cleaned datasets
print("CLEANED STUDENT DATA")
display(clean_students)

print("CLEANED ATTENDANCE DATA")
display(clean_attendance)

# Final cleaning summary
print("\nFINAL DATA CLEANING SUMMARY")
print("Total Students:", len(clean_students))
print("Total Attendance Records:", len(clean_attendance))

print("Missing Values in Student Data:",
      clean_students.isnull().sum().sum())

print("Missing Values in Attendance Data:",
      clean_attendance.isnull().sum().sum())

# Save cleaned datasets
clean_students.to_csv(
    "Cleaned_Student_Data.csv", index=False
)

clean_attendance.to_csv(
    "Cleaned_Attendance_Data.csv", index=False
)

print("\nCleaned datasets saved successfully!")
```

##Conclusion
In this task, final data cleaning was performed on the Student Attendance Management System datasets. Missing values in essential fields were handled, duplicate records were removed, and data types were corrected to maintain consistency. Invalid attendance statuses and records belonging to unregistered students were also removed.

The cleaned datasets were checked and saved as separate CSV files for future use. This process improves the accuracy, reliability, and quality of the attendance information. Final data cleaning is an important step before preparing the project documentation and completing the system.

```python

# ============================================================
# STUDENT ATTENDANCE MANAGEMENT SYSTEM
# YBI FOUNDATION - PYTHON PROGRAMMING INTERNSHIP
# COMPLETE PROJECT: DAY 1 TO DAY 45
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import zipfile

from IPython.display import display

print("=" * 65)
print("STUDENT ATTENDANCE MANAGEMENT SYSTEM")
print("YBI FOUNDATION PYTHON PROGRAMMING INTERNSHIP")
print("=" * 65)


# 2. CREATE STUDENT DATASET
student_data = {
    "Student_ID": [101, 102, 103, 104, 105,
                   106, 107, 108, 109, 110],

    "Name": ["Aarav", "Riya", "Rahul", "Priya", "Ankit",
             "Sneha", "Aditya", "Neha", "Rohan", "Ananya"],

    "Branch": ["BCA"] * 10,
    "Semester": [3] * 10,
    "Section": ["A"] * 10
}

students = pd.DataFrame(student_data)

print("\n1. STUDENT DATASET")
display(students)


# 3. CREATE ATTENDANCE DATASET
attendance_data = {
    "Student_ID": list(range(101, 111)) +
                  list(range(101, 111)),

    "Date": ["2026-09-01"] * 10 +
            ["2026-09-02"] * 10,

    "Subject": ["Python"] * 10 +
               ["DBMS"] * 10,

    "Status": [
        "Present", "Present", "Absent", "Present", "Present",
        "Absent", "Present", "Present", "Absent", "Present",

        "Present", "Absent", "Present", "Present", "Absent",
        "Present", "Present", "Absent", "Present", "Present"
    ]
}

attendance_data = pd.DataFrame(attendance_data)

print("\n2. ATTENDANCE DATASET")
display(attendance_data)


# 4. COMBINE STUDENT AND ATTENDANCE DATA
combined_data = pd.merge(
    attendance_data,
    students,
    on="Student_ID",
    how="left"
)

print("\n3. COMBINED DATASET")
display(combined_data)


# 5. SAVE AND LOAD CSV DATA
students.to_csv("Student_Data.csv", index=False)
attendance_data.to_csv("Attendance_Data.csv", index=False)

students = pd.read_csv("Student_Data.csv")
attendance_data = pd.read_csv("Attendance_Data.csv")

attendance_data["Date"] = pd.to_datetime(
    attendance_data["Date"]
)

print("\n4. CSV FILES CREATED AND LOADED SUCCESSFULLY")


# 6. DATASET EXPLORATION
print("\n5. DATASET INFORMATION")
print("Student Dataset Shape:", students.shape)
print("Attendance Dataset Shape:", attendance_data.shape)

print("\nStudent Dataset Information:")
students.info()

print("\nAttendance Dataset Information:")
attendance_data.info()

print("\nAttendance Statistics:")
print(attendance_data.describe(include="all"))


# 7. MISSING VALUE ANALYSIS
print("\n6. MISSING VALUE ANALYSIS")
print("Student Missing Values:")
print(students.isnull().sum())

print("\nAttendance Missing Values:")
print(attendance_data.isnull().sum())


# 8. REMOVE DUPLICATES
students = students.drop_duplicates(
    subset="Student_ID"
).copy()

attendance_data = attendance_data.drop_duplicates().copy()

print("\n7. DUPLICATE RECORDS REMOVED")


# 9. FINAL DATA CLEANING
students = students.dropna(
    subset=["Student_ID", "Name"]
).copy()

attendance_data = attendance_data.dropna(
    subset=["Student_ID", "Date", "Subject", "Status"]
).copy()

attendance_data["Status"] = (
    attendance_data["Status"]
    .astype(str)
    .str.strip()
    .str.title()
)

attendance_data = attendance_data[
    attendance_data["Status"].isin(["Present", "Absent"])
].copy()

attendance_data = attendance_data[
    attendance_data["Student_ID"].isin(students["Student_ID"])
].copy()

students = students.reset_index(drop=True)
attendance_data = attendance_data.reset_index(drop=True)

print("\n8. CLEANED STUDENT DATA")
display(students)

print("\nCLEANED ATTENDANCE DATA")
display(attendance_data)


# 10. TOTAL, PRESENT AND ABSENT CLASSES
total_classes = attendance_data.groupby(
    "Student_ID"
).size()

present_classes = (
    attendance_data[
        attendance_data["Status"] == "Present"
    ]
    .groupby("Student_ID")
    .size()
)

attendance_summary = pd.DataFrame({
    "Total_Classes": total_classes,
    "Present_Classes": present_classes
}).fillna(0)

attendance_summary["Present_Classes"] = (
    attendance_summary["Present_Classes"].astype(int)
)

attendance_summary["Absent_Classes"] = (
    attendance_summary["Total_Classes"] -
    attendance_summary["Present_Classes"]
)

attendance_summary = attendance_summary.reset_index()


# 11. CALCULATE ATTENDANCE PERCENTAGE
attendance_summary["Attendance_Percentage"] = (
    attendance_summary["Present_Classes"] /
    attendance_summary["Total_Classes"] * 100
).round(2)

print("\n9. ATTENDANCE SUMMARY")
display(attendance_summary)


# 12. ATTENDANCE STATUS
attendance_summary["Attendance_Status"] = (
    attendance_summary["Attendance_Percentage"].apply(
        lambda x: "Good Attendance" if x >= 75
        else "Low Attendance"
    )
)


# 13. STUDENT-WISE ATTENDANCE REPORT
attendance_report = pd.merge(
    students,
    attendance_summary,
    on="Student_ID",
    how="left"
)

print("\n10. STUDENT-WISE ATTENDANCE REPORT")
display(attendance_report)


# 14. SUBJECT-WISE ANALYSIS
subject_summary = attendance_data.groupby(
    "Subject"
).agg(
    Total_Classes=("Status", "count"),
    Present_Classes=("Status", lambda x: (x == "Present").sum()),
    Absent_Classes=("Status", lambda x: (x == "Absent").sum())
).reset_index()

subject_summary["Attendance_Percentage"] = (
    subject_summary["Present_Classes"] /
    subject_summary["Total_Classes"] * 100
).round(2)

print("\n11. SUBJECT-WISE ATTENDANCE")
display(subject_summary)


# 15. DATE-WISE ANALYSIS
date_summary = attendance_data.groupby(
    "Date"
).agg(
    Total_Records=("Status", "count"),
    Present_Students=("Status", lambda x: (x == "Present").sum()),
    Absent_Students=("Status", lambda x: (x == "Absent").sum())
).reset_index()

date_summary["Attendance_Percentage"] = (
    date_summary["Present_Students"] /
    date_summary["Total_Records"] * 100
).round(2)

print("\n12. DATE-WISE ATTENDANCE")
display(date_summary)


# 16. BRANCH-WISE ANALYSIS
branch_data = pd.merge(
    attendance_data,
    students[["Student_ID", "Branch"]],
    on="Student_ID",
    how="left"
)

branch_summary = branch_data.groupby(
    "Branch"
).agg(
    Total_Records=("Status", "count"),
    Present_Students=("Status", lambda x: (x == "Present").sum()),
    Absent_Students=("Status", lambda x: (x == "Absent").sum())
).reset_index()

branch_summary["Attendance_Percentage"] = (
    branch_summary["Present_Students"] /
    branch_summary["Total_Records"] * 100
).round(2)

print("\n13. BRANCH-WISE ATTENDANCE")
display(branch_summary)


# 17. LOW ATTENDANCE STUDENTS
low_attendance = attendance_report[
    attendance_report["Attendance_Percentage"] < 75
]

print("\n14. LOW ATTENDANCE STUDENTS")
display(low_attendance)


# 18. GOOD ATTENDANCE STUDENTS
good_attendance = attendance_report[
    attendance_report["Attendance_Percentage"] >= 75
]

print("\n15. GOOD ATTENDANCE STUDENTS")
display(good_attendance)


# 19. HIGHEST ATTENDANCE STUDENTS
highest_attendance = attendance_report[
    attendance_report["Attendance_Percentage"] ==
    attendance_report["Attendance_Percentage"].max()
]

print("\n16. HIGHEST ATTENDANCE STUDENTS")
display(highest_attendance)


# 20. ATTENDANCE STATISTICS
overall_attendance = (
    attendance_report["Present_Classes"].sum() /
    attendance_report["Total_Classes"].sum() * 100
)

print("\n17. ATTENDANCE STATISTICS")
print("Total Students:", len(students))
print("Total Attendance Records:", len(attendance_data))
print("Total Present:", attendance_data["Status"].eq("Present").sum())
print("Total Absent:", attendance_data["Status"].eq("Absent").sum())
print("Overall Attendance:", round(overall_attendance, 2), "%")
print("Average Student Attendance:",
      round(attendance_report["Attendance_Percentage"].mean(), 2), "%")
print("Highest Attendance:",
      attendance_report["Attendance_Percentage"].max(), "%")
print("Lowest Attendance:",
      attendance_report["Attendance_Percentage"].min(), "%")


# 21. PRESENT VS ABSENT VISUALIZATION
status_counts = attendance_data["Status"].value_counts()

plt.figure(figsize=(7, 5))
status_counts.reindex(["Present", "Absent"], fill_value=0).plot(
    kind="bar"
)
plt.title("Present vs Absent Students")
plt.xlabel("Attendance Status")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# 22. ATTENDANCE DISTRIBUTION PIE CHART
plt.figure(figsize=(6, 6))
status_counts.reindex(["Present", "Absent"], fill_value=0).plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Attendance Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()


# 23. STUDENT-WISE ATTENDANCE CHART
plt.figure(figsize=(10, 5))
plt.bar(
    attendance_report["Name"],
    attendance_report["Attendance_Percentage"]
)
plt.axhline(
    y=75,
    linestyle="--",
    label="75% Attendance Threshold"
)
plt.title("Student-Wise Attendance Percentage")
plt.xlabel("Students")
plt.ylabel("Attendance Percentage")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()


# 24. SUBJECT-WISE ATTENDANCE CHART
plt.figure(figsize=(7, 5))
plt.bar(
    subject_summary["Subject"],
    subject_summary["Attendance_Percentage"]
)
plt.title("Subject-Wise Attendance Percentage")
plt.xlabel("Subject")
plt.ylabel("Attendance Percentage")
plt.tight_layout()
plt.show()


# 25. BRANCH-WISE ATTENDANCE CHART
plt.figure(figsize=(7, 5))
plt.bar(
    branch_summary["Branch"],
    branch_summary["Attendance_Percentage"]
)
plt.title("Branch-Wise Attendance Percentage")
plt.xlabel("Branch")
plt.ylabel("Attendance Percentage")
plt.tight_layout()
plt.show()


# 26. ATTENDANCE PERCENTAGE DISTRIBUTION
plt.figure(figsize=(8, 5))
plt.hist(
    attendance_report["Attendance_Percentage"],
    bins=5,
    edgecolor="black"
)
plt.title("Attendance Percentage Distribution")
plt.xlabel("Attendance Percentage")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()


# 27. MONTHLY ATTENDANCE ANALYSIS
attendance_data["Month"] = (
    attendance_data["Date"].dt.to_period("M").astype(str)
)

monthly_summary = attendance_data.groupby(
    "Month"
).agg(
    Total_Records=("Status", "count"),
    Present_Records=("Status", lambda x: (x == "Present").sum()),
    Absent_Records=("Status", lambda x: (x == "Absent").sum())
).reset_index()

monthly_summary["Attendance_Percentage"] = (
    monthly_summary["Present_Records"] /
    monthly_summary["Total_Records"] * 100
).round(2)

print("\n18. MONTHLY ATTENDANCE ANALYSIS")
display(monthly_summary)


# 28. SUMMARY TABLE
summary_table = attendance_report[
    ["Student_ID", "Name", "Total_Classes",
     "Present_Classes", "Absent_Classes",
     "Attendance_Percentage", "Attendance_Status"]
].copy()

print("\n19. FINAL SUMMARY TABLE")
display(summary_table)


# 29. SEARCH STUDENT ATTENDANCE
search_id = 103

searched_student = attendance_report[
    attendance_report["Student_ID"] == search_id
]

print("\n20. STUDENT SEARCH RESULT")
if not searched_student.empty:
    display(searched_student)
else:
    print("Student not found.")


# 30. INDIVIDUAL STUDENT REPORT
individual_report = attendance_report[
    attendance_report["Student_ID"] == search_id
].copy()

print("\n21. INDIVIDUAL STUDENT REPORT")
display(individual_report)


# 31. OVERALL ATTENDANCE REPORT
overall_report = pd.DataFrame({
    "Metric": [
        "Total Students",
        "Total Attendance Records",
        "Present Records",
        "Absent Records",
        "Overall Attendance Percentage",
        "Good Attendance Students",
        "Low Attendance Students"
    ],

    "Value": [
        len(students),
        len(attendance_data),
        int((attendance_data["Status"] == "Present").sum()),
        int((attendance_data["Status"] == "Absent").sum()),
        round(overall_attendance, 2),
        len(good_attendance),
        len(low_attendance)
    ]
})

print("\n22. OVERALL ATTENDANCE REPORT")
display(overall_report)


# 32. ATTENDANCE RECOMMENDATIONS
attendance_recommendations = attendance_report[
    ["Student_ID", "Name", "Attendance_Percentage"]
].copy()

attendance_recommendations["Recommendation"] = (
    attendance_recommendations["Attendance_Percentage"].apply(
        lambda x: "Improve attendance regularly"
        if x < 75 else "Maintain good attendance"
    )
)

print("\n23. ATTENDANCE RECOMMENDATIONS")
display(attendance_recommendations)


# 33. PROJECT INSIGHTS
print("\n24. PROJECT INSIGHTS")

print("1. Total students analyzed:", len(students))
print("2. Average attendance:",
      round(attendance_report["Attendance_Percentage"].mean(), 2), "%")
print("3. Students with good attendance:", len(good_attendance))
print("4. Students with low attendance:", len(low_attendance))
print("5. Overall attendance:", round(overall_attendance, 2), "%")


# 34. PROJECT RECOMMENDATIONS
print("\n25. PROJECT RECOMMENDATIONS")

print("- Students below 75% should improve attendance.")
print("- Regular attendance records should be maintained.")
print("- Attendance reports should be reviewed regularly.")
print("- Student attendance should be monitored continuously.")


# 35. FINAL DATA QUALITY CHECK
print("\n26. FINAL DATA QUALITY CHECK")

print("Student Missing Values:",
      students.isnull().sum().sum())

print("Attendance Missing Values:",
      attendance_data.drop(columns=["Month"]).isnull().sum().sum())

print("Duplicate Student Records:",
      students.duplicated(subset=["Student_ID"]).sum())

print("Duplicate Attendance Records:",
      attendance_data.drop(columns=["Month"]).duplicated().sum())


# 36. SAVE ALL PROJECT REPORTS
students.to_csv("Cleaned_Student_Data.csv", index=False)

attendance_data.drop(
    columns=["Month"]
).to_csv("Cleaned_Attendance_Data.csv", index=False)

combined_data.to_csv(
    "Combined_Attendance_Data.csv", index=False
)

attendance_report.to_csv(
    "Student_Attendance_Report.csv", index=False
)

overall_report.to_csv(
    "Overall_Student_Attendance_Report.csv", index=False
)

attendance_recommendations.to_csv(
    "Attendance_Recommendations.csv", index=False
)

subject_summary.to_csv(
    "Subject_Wise_Report.csv", index=False
)

date_summary.to_csv(
    "Date_Wise_Report.csv", index=False
)

branch_summary.to_csv(
    "Branch_Wise_Report.csv", index=False
)

monthly_summary.to_csv(
    "Monthly_Attendance_Report.csv", index=False
)

low_attendance.to_csv(
    "Low_Attendance_Students.csv", index=False
)

good_attendance.to_csv(
    "Good_Attendance_Students.csv", index=False
)

individual_report.to_csv(
    "Individual_Student_Report.csv", index=False
)

# 38. CREATE README FILE
readme = """
# Student Attendance Management System

## Project Overview
A Python-based project for analyzing student attendance records.

## Technologies Used
- Python
- Pandas
- NumPy
- Matplotlib
- Google Colab
- GitHub

## Main Features
- Student dataset creation
- Attendance management
- Data cleaning
- Attendance percentage calculation
- Student-wise analysis
- Subject-wise and branch-wise analysis
- Attendance visualizations
- Attendance reports
- Student search
- Attendance recommendations

## How to Run
1. Open Google Colab.
2. Upload the project notebook.
3. Run the notebook cells.
4. View the generated reports and charts.

## Internship
YBI Foundation Python Programming Internship
"""

with open(
    "README.md", "w", encoding="utf-8"
) as file:
    file.write(readme)


# 39. FINAL COMPLETION REPORT
with open(
    "Final_Project_Completion_Report.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write("STUDENT ATTENDANCE MANAGEMENT SYSTEM\n")
    file.write("=" * 45 + "\n")
    file.write("Project Duration: 45 Days\n")
    file.write("Internship: YBI Foundation\n")
    file.write("Total Students: " + str(len(students)) + "\n")
    file.write("Total Attendance Records: " +
               str(len(attendance_data)) + "\n")
    file.write("Overall Attendance: " +
               str(round(overall_attendance, 2)) + "%\n")
    file.write("Project Status: Completed\n")



# 39. FINAL COMPLETION REPORT
with open(
    "Final_Project_Completion_Report.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write("STUDENT ATTENDANCE MANAGEMENT SYSTEM\n")
    file.write("=" * 45 + "\n")
    file.write("Project Duration: 45 Days\n")
    file.write("Internship: YBI Foundation\n")
    file.write("Total Students: " + str(len(students)) + "\n")
    file.write("Total Attendance Records: " +
               str(len(attendance_data)) + "\n")
    file.write("Overall Attendance: " +
               str(round(overall_attendance, 2)) + "%\n")
    file.write("Project Status: Completed\n")


# 40. CREATE PROJECT FILE LIST
project_files = [
    "Student_Data.csv",
    "Attendance_Data.csv",
    "Combined_Attendance_Data.csv",
    "Cleaned_Student_Data.csv",
    "Cleaned_Attendance_Data.csv",
    "Student_Attendance_Report.csv",
    "Overall_Student_Attendance_Report.csv",
    "Attendance_Recommendations.csv",
    "Subject_Wise_Report.csv",
    "Date_Wise_Report.csv",
    "Branch_Wise_Report.csv",
    "Monthly_Attendance_Report.csv",
    "Low_Attendance_Students.csv",
    "Good_Attendance_Students.csv",
    "Individual_Student_Report.csv",
    "Project_Documentation.txt",
    "README.md",
    "Final_Project_Completion_Report.txt"
]

print("\n27. FINAL PROJECT FILES")

for filename in project_files:
    print(filename, "-", "Available" if os.path.exists(filename) else "Missing")


# 41. FINAL PROJECT VALIDATION
assert len(students) == 10
assert len(attendance_data) == 20
assert (attendance_data["Status"] == "Present").sum() == 14
assert (attendance_data["Status"] == "Absent").sum() == 6
assert attendance_data.isnull().sum().sum() == 0

print("\nALL PROJECT VALIDATION TESTS PASSED!")


# 42. CREATE ZIP FILE
with zipfile.ZipFile(
    "Student_Attendance_Management_System.zip",
    "w",
    zipfile.ZIP_DEFLATED
) as zip_file:

    for filename in project_files:
        if os.path.exists(filename):
            zip_file.write(filename)

print("\nPROJECT ZIP FILE CREATED SUCCESSFULLY!")


# 43. FINAL OUTPUT
print("\n" + "=" * 65)
print("STUDENT ATTENDANCE MANAGEMENT SYSTEM")
print("45-DAY PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 65)

print("Total Students:", len(students))
print("Total Attendance Records:", len(attendance_data))
print("Present Records:", present_count if "present_count" in globals()
      else int((attendance_data["Status"] == "Present").sum()))
print("Absent Records:", absent_count if "absent_count" in globals()
      else int((attendance_data["Status"] == "Absent").sum()))
print("Overall Attendance:", round(overall_attendance, 2), "%")

print("\nGenerated Project File:")
print("Student_Attendance_Management_System.zip")

print("\nDownload the ZIP file from the Colab Files panel.")
print("FINAL PROJECT EXECUTION COMPLETED!")
```
