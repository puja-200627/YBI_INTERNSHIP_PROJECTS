# Day 5 - Combining Student and Attendance Data

In this step, the student information and attendance records are combined using the Student_ID column.

The combined dataset will contain both student details and attendance information.

##Merging Both Datasets

```python
import pandas as pd
student_data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Puja", "Riya", "Rahul", "Priya", "Ankit",
        "Sneha", "Aditya", "Neha", "Rohan", "Ananya"
    ],
    "Branch": [
        "BCA", "BCA", "BCA", "BCA", "BCA",
        "BCA", "BCA", "BCA", "BCA", "BCA"
    ],
    "Semester": [3, 3, 3, 3, 3, 3, 3, 3, 3, 3],
    "Section": ["A", "A", "A", "A", "A", "A", "A", "A", "A", "A"]
}
attendance_data = {
    "Student_ID": [
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110,
        101, 102, 103, 104, 105,
        106, 107, 108, 109, 110
    ],

    "Date": [
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01", "2026-09-01",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02",
        "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02", "2026-09-02"
    ],

    "Subject": [
        "Python", "Python", "Python", "Python", "Python",
        "Python", "Python", "Python", "Python", "Python",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS"
    ],

    "Status": [
        "Present", "Present", "Absent", "Present", "Present",
        "Absent", "Present", "Present", "Absent", "Present",
        "Present", "Absent", "Present", "Present", "Absent",
        "Present", "Present", "Absent", "Present", "Present"
    ]
}
attendance = pd.DataFrame(attendance_data)
students = pd.DataFrame(student_data)
combined_data = pd.merge(
    attendance,
    students,
    on="Student_ID",
    how="left"
)

combined_data
```
